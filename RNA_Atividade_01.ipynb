Crie um modelo perceptron e aplique para prever os seguintes casos:
1. Caso do AND
2. Caso do OR
3. Caso do XOR

Para cada caso defina sua própria estratégia de inicialização dos pesos. Comente os resultados obtidos em cada caso.

Importações e Funções

```python
import numpy as np
import matplotlib.pyplot as plt

# Dados de entrada comuns para todos os casos e possibilidades
entradas = np.array([[0,0],[0,1],[1,0],[1,1]])

taxa_aprendizado = 0.1
epocas = 100

# Função de ativação degrau (transforma as entradas na saída)
def funcao_ativacao(x):
    return 1 if x >= 0 else 0

# Função de plotagem da fronteira de decisão
def plotar_fronteira_decisao(pesos, bias, entradas, saidas, epoca, titulo):
    plt.clf()  # Limpa a figura atual para cada nova plotagem

    # Plota os pontos de dados
    plt.scatter(entradas[:, 0], entradas[:, 1], c=saidas.flatten(), cmap=plt.cm.coolwarm, s=100)

    # Plota a linha de decisão (fronteira)
    if pesos[1] != 0:
        x_vals = np.array([0, 1])
        y_vals = (-pesos[0] / pesos[1]) * x_vals - (bias / pesos[1])
        plt.plot(x_vals, y_vals, 'k--')

    plt.title(f"{titulo} - Época {epoca + 1}")
    plt.xlim(-0.5, 1.5)
    plt.ylim(-0.5, 1.5)
    plt.grid(True)
    plt.show(block=False)
    plt.pause(0.1)  # Pausa para ver o gráfico atualizado
```

Caso AND

```python

saidas_e = np.array([[0],[0],[0],[1]])

# Inicialização dos pesos e bias
np.random.seed(42)
pesos_and = np.random.randn(2)
bias_and = np.random.randn(1)

print(f"Pesos iniciais AND: {pesos_and}")
print(f"Bias Inicial AND: {bias_and}")

# Loop de treinamento para AND
print("--- Iniciando o treinamento AND ---")
for epoca in range(epocas):
    erro_total = 0
    for i in range(len(entradas)):
        soma_ponderada = np.dot(entradas[i], pesos_and) + bias_and
        predicao = funcao_ativacao(soma_ponderada)
        erro = saidas_e[i] - predicao
        erro_total += abs(erro)

        if erro != 0:
            pesos_and = pesos_and + taxa_aprendizado * erro * entradas[i]
            bias_and = bias_and + taxa_aprendizado * erro

    print(f"Época {epoca + 1}: Pesos={pesos_and}, Bias={bias_and}, Erro total={erro_total}")

    # Chamada para a função de plotagem
    plotar_fronteira_decisao(pesos_and, bias_and, entradas, saidas_e, epoca, "AND")

    if erro_total == 0:
        print("\nTreinamento completo AND! Modelo convergiu.")
        break

print("\n--- Treinamento finalizado AND ---")
print(f"Pesos finais AND: {pesos_and}")
print(f"Bias final AND: {bias_and}")

# Testando o modelo final
print("\n--- Teste final do modelo AND ---")
for i in range(len(entradas)):
    soma_ponderada = np.dot(entradas[i], pesos_and) + bias_and
    predicao = funcao_ativacao(soma_ponderada)
    print(f"Entrada: {entradas[i]}, Previsão: {predicao}, Saída esperada: {saidas_e[i]}")
```


O modelo conseguiu aprender. O treinamento convergiu rapidamente porque o problema do AND é linearmente separável. Em outras palavras, é possível traçar uma linha reta que separa perfeitamente a saída 1 das saídas 0.



Case OR

O modelo também conseguiu aprender. Ele resolveu o problema porque a lógica OR, assim como a AND, é linearmente separável. A linha de separação é diferente, mas o perceptron foi capaz de encontrá-la e convergir.
